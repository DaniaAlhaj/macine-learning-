1)Regression tasks:1.Apply linear regression to learn the attribute “Customer Value” using all independent attributes (call this model LRM1) 

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score
import matplotlib.pyplot as plt
import numpy as np

data = pd.read_csv('/home/dania-alhaj/machin+assignment/machin assignment/CustomerChurn.csv')
data['Churn'] = data['Churn'].map({'yes': 1, 'no': 0})
data['Complains'] = data['Complains'].map({'yes': 1, 'no': 0})
data['Status'] = data['Status'].map({'active': 1, 'not-active': 0})
data['Plan'] = data['Plan'].map({'pre-paid': 1, 'post-paid': 0})

X = data[['Call Failure', 'Complains', 'Charge Amount', 'Freq. of SMS', 'Distinct Called Numbers',
          'Age Group', 'Age', 'Plan', 'Status', 'Freq. of use', 'Churn']]
y = data['Customer Value']
X_train, X_test, y_train, y_test= train_test_split(X, y, test_size=0.3, random_state=42)

LRM1 = LinearRegression()
LRM1.fit(X_train, y_train)

y_pred_train = LRM1.predict(X_train)

mse_train = mean_squared_error(y_train, y_pred_train)
rmse_train = np.sqrt(mse_train)
mae_train = mean_absolute_error(y_train, y_pred_train)
r2_train = r2_score(y_train, y_pred_train)

print("Training Metrics:")
print(f"R-squared (R2): {r2_train:.2f}")
print(f"Mean Squared Error (MSE): {mse_train:.2f}")
print(f"Root Mean Squared Error (RMSE): {rmse_train:.2f}")
print(f"Mean Absolute Error (MAE): {mae_train:.2f}")

plt.scatter(y_train, y_pred_train, color='blue', alpha=0.6)
plt.plot([y_train.min(), y_train.max()], [y_train.min(), y_train.max()], color='red')
plt.xlabel("Actual Customer Value")
plt.ylabel("Predicted Customer Value")
plt.title("Actual vs Predicted Customer Value (Training Data Only)")
plt.show()
```

adequate performance metrics for LRM1 training data  , R-squared = 0.97:indicates that 97% of the variability in the customer value is explained by the selected features in the training data so the model fits the training data very well. 
Root Mean Squared Error (RMSE): 91.76 its the average prediction error is nearly 91.76 units so we have a low error (here is a reasonable amount of error in the predictions )
Mean Absolute Error = 60.60: as we know its the average absolute difference between actual and predicted values.the model's predictions are 60.60 units off from the true customer value. as we know MAE is the average absolute difference between the predicted and actual customer values.
